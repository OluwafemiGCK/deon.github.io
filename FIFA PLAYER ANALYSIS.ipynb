```python
from sklearn import svm
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier, export_graphviz
from sklearn import metrics
```

```python
#Extracting only the independent variables
X = fifa.iloc[:, 8].values #Potential, Special

#Extracting only the dependent variables
y = fifa.iloc[:, 7].values #Real Face
```

```python
# Splitting the dataset into Training Set and Test Set
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=4) 
```

```python
X_train_mod = X_train.reshape(-1,1)
```

```python
X_test_mod = X_test.reshape(-1,1)
```

```python
y_train_mod = y_train.reshape(-1,1)
```

```python
y_test_mod = y_test.reshape(-1,1)
```

```python
# Fitting SVM to trainig dataset
model = svm.SVC(kernel = 'linear')
```

```python
model.fit(X_train_mod, y_train_mod)
```

```python
fifa.iloc[:, 7]
```

```python
y_pred_mod = model.predict(X_test_mod)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy = accuracy_score(y_test_mod, y_pred_mod)
print(accuracy)
```
