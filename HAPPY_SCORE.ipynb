# USING K-MEANS TO INTERPRET THE DATA

```python
from sklearn.cluster import KMeans

import numpy as np

income_happy = np.column_stack((income, happy))

km_res = KMeans(n_clusters=2).fit(income_happy)

clusters = km_res.cluster_centers_

plt.scatter(income, happy)
plt.scatter(clusters[:,0], clusters[:,1], s=1000, alpha=0.75)
```

# BUILDING A MODEL

Building a BINARY CLASSIFICATION MODEL

```python
v1_v2 = np.column_stack((data2.V1, data2.V2))

#Compute K-Means

km_res = KMeans(n_clusters = 2).fit(v1_v2)
clusters = km_res.cluster_centers_

data2['KMeans'] = km_res.labels_



plt.scatter(x = data2.V1, y = data2.V2)
plt.scatter(clusters[:, 0], clusters[:, 1], s=500, marker = '*', c='r', alpha=0.25)
```
